```python
print('Enter your choice : ')
print('1. Create ;')
print('2. Update ;')
print('3. Read ;')
print('4. Delete ;')
ip = int(input())

```

```python
if (ip == 1):
  id = input("Enter id :")
  name = input("Enter Name :")
  department = input("Enter Department :")
  salary = input("Enter salary :")
  str1 = id + name + department + salary + "\n"
  file = open("data.csv","w")
  data = file.write(str1)
  file.close()
  print('Data added successfully;')

elif(ip == 2):
  up_id = input("Enter emp_id to update : ")
  up_data = input("Enter the data to update : ")+"\n"

  with open("data.txt", "r") as file:
    lines = file.readlines()

  update_lines = []
  for line in lines:
    if up_id in line:
      update_lines.append(up_data)
    else :
      update_lines.append(line)

  with open("data.csv", "w") as file:
    file.writelines(update_lines)

  print('Data updated successfully;')

elif(ip == 3):
  with open("data.csv", mode="r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(
            f"Emp_ID: {row['Emp_ID']}\n"
            f"Name: {row['Name']}\n"
            f"Department: {row['Department']}\n"
            f"Salary: {row['Salary']}\n"
            f"{'-' * 20}"
        )

elif(ip == 4):

  del_id = input("Enter emp_id to delete : ")
  with open("data.csv", "r") as file:
    lines = file.readlines()

  delete_lines = []
  for line in lines:
    if del_id not in line:
      delete_lines.append(line)

  with open("data.csv", "w") as file:
    file.writelines(delete_lines)

  print('Deleted Sucessfully')

else :
  print("Invalid Operation!!!")
```
