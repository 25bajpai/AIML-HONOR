```python
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="your_password",
    database="employee_db",
)

cursor = db.cursor()
```

```python
x = int(input("Enter you choice : \n1. Create \n2.Read \n3.Update \n4.Delete"))
```

```python
# Create
if x equals 1:
  insert_query = "INSERT INTO tblemployee (empname, department, salary) VALUES (%s, %s, %s)"
  employee_data = ("Aarav Sharma", "Engineering", 75000)

  cursor.execute(insert_query, employee_data)
  db.commit()
  print("Record inserted successfully. ID:", cursor.lastrowid)
```

```python
# Read
if x equals 2:
  read_query = "SELECT empid, empname, department, salary FROM tblemployee"
  cursor.execute(read_query)

  rows = cursor.fetchall()
  print("\n Employee Records ---")
  for row in rows:
    print(row)
```

```python
# Update
if x equals 3:

  update_query = "UPDATE tblemployee SET salary = %s WHERE empid = %s"
  update_data = (82000, 1)  # Setting salary to 82000 for empid = 1

  cursor.execute(update_query, update_data)
  db.commit()
  print("\nRecord updated successfully.")
```

```python
# Delete
if x equals 4:
  delete_query = "DELETE FROM tblemployee WHERE empid = %s"
  delete_data = (1,)

  cursor.execute(delete_query, delete_data)
  db.commit()
  print("Record deleted successfully.")

  cursor.close()
  db.close()
```
